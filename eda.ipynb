```python
import pandas as pd
from pathlib import Path

data_folder = Path(r"C:\Users\Admin\Desktop\Module5\Data")

output_folder = data_folder / "Cleaned"
output_folder.mkdir(exist_ok=True)


files = [
    "03_Library Systembook.csv",
    "03_Library SystemCustomers.csv"
]

for file_name in files:

    file_path = data_folder / file_name

    df = pd.read_csv(file_path)

    print(f"Cleaning: {file_name}")
    print(f"Rows before cleaning: {len(df)}")

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )


    for column in df.select_dtypes(include="object").columns:
        df[column] = df[column].str.strip()

    df = df.replace(r"^\s*$", pd.NA, regex=True)

    df = df.dropna(how="all")

    df = df.drop_duplicates()

    print(f"Rows after cleaning: {len(df)}")

    output_name = file_path.stem + "_clean.csv"
    output_path = output_folder / output_name

    df.to_csv(output_path, index=False)

    print(f"Saved to: {output_path}")
    print("-" * 50)

print("Cleansed data outputted")
```
